```python
import pandas as pd

# Criando uma série simples
series_objeto = pd.Series([1, 7, 9, 13, 17, 99])
print("Série Simples:")
display(series_objeto) # 'display' deixa a tabela bonita no VS Code

# Criando série com índices personalizados
series_objeto2 = pd.Series([4, 7, 8, -2], index=['alfa', 'beta', 'teta', 'gama'])
print("\nSérie Personalizada:")
display(series_objeto2)
```

```python
# Criando as listas iniciais
cidades = ['Maringá', 'Itabira', 'Uberlândia', 'Salvador', 'Fortaleza']
populacao = [403063, 120904, 699097, 2886698, 2686612]

# Transformando em Dicionário (zip)
populacao_cidades = dict(zip(cidades, populacao))

print("Dicionário Criado:")
display(populacao_cidades)

# Acessando valor específico
print(f"\nPopulação de Uberlândia: {populacao_cidades['Uberlândia']}")

# Verificando se existe
print(f"Maringá existe? {'Maringá' in populacao_cidades}")
print(f"São Carlos existe? {'São Carlos' in populacao_cidades}")

# Adicionando Vitória
populacao_cidades['Vitória'] = 365855

# Deletando Fortaleza
del populacao_cidades['Fortaleza']

print("\nDicionário Final (Com Vitória, Sem Fortaleza):")
display(populacao_cidades)
```

```python
# Criando os dados
dados = {
    'Autor': ['Sun Tzu', 'Fernando Pessoa', 'Thomas Mann', 'João Guimarães Rosa'],
    'Livro': ['A Arte da Guerra', 'Poesias Selecionadas', 'A Montanha Mágica', 'Primeiras Estórias'],
    'Ano': [2000, 2004, 2015, 2017]
}

# Criando o DataFrame
autores = pd.DataFrame(dados)

# Verificando o tipo
print(type(autores))

# Salvando em arquivo CSV (Vai aparecer na pasta do seu projeto)
autores.to_csv("autores.csv")

print("Arquivo autores.csv criado com sucesso!")
```

```python
# Lendo o arquivo que acabamos de criar
# Note que não usamos '/content/...', apenas o nome do arquivo pois está na mesma pasta
df_leitura = pd.read_csv('autores.csv', index_col=0)

print("--- HEAD (Primeiras linhas) ---")
display(df_leitura.head())

print("\n--- DESCRIBE (Estatísticas) ---")
display(df_leitura.describe())

print("\n--- INFO (Informações Técnicas) ---")
df_leitura.info()

print("\n--- COLUNAS ---")
print(df_leitura.columns)

print("\n--- ÍNDICES ---")
print(df_leitura.index)
```
